## 8. Deterministic Tests

Run this cell after implementing every TODO. The tests cover the four operations, boundaries, the cross and 5x5 elements, output contract, and input immutability.

```python
def assert_binary_output(output, shape):
    assert output.shape == shape
    assert np.issubdtype(output.dtype, np.integer)
    assert set(np.unique(output)).issubset({0, 255})


single_pixel = np.zeros((7, 7), dtype=np.uint8)
single_pixel[3, 3] = 255
expected_dilation = np.zeros((7, 7), dtype=np.uint8)
expected_dilation[2:5, 2:5] = 255
np.testing.assert_array_equal(
    dilatacao_binaria(single_pixel, square_3), expected_dilation
)

block = np.zeros((7, 7), dtype=np.uint8)
block[2:5, 2:5] = 255
expected_erosion = np.zeros((7, 7), dtype=np.uint8)
expected_erosion[3, 3] = 255
np.testing.assert_array_equal(erosao_binaria(block, square_3), expected_erosion)

with_noise = block.copy()
with_noise[0, 0] = 255
np.testing.assert_array_equal(abertura_binaria(with_noise, square_3), block)

with_hole = np.zeros((9, 9), dtype=np.uint8)
with_hole[2:7, 2:7] = 255
with_hole[4, 4] = 0
expected_closing = with_hole.copy()
expected_closing[4, 4] = 255
np.testing.assert_array_equal(fechamento_binario(with_hole, square_3), expected_closing)

border_pixel = np.zeros((5, 5), dtype=np.uint8)
border_pixel[0, 0] = 255
expected_border = np.zeros((5, 5), dtype=np.uint8)
expected_border[0:2, 0:2] = 255
np.testing.assert_array_equal(
    dilatacao_binaria(border_pixel, square_3), expected_border
)

expected_cross = np.zeros((7, 7), dtype=np.uint8)
expected_cross[2:5, 3] = 255
expected_cross[3, 2:5] = 255
np.testing.assert_array_equal(dilatacao_binaria(single_pixel, cross_3), expected_cross)

large_image = np.zeros((9, 9), dtype=np.uint8)
large_image[4, 4] = 255
expected_square_5 = np.zeros((9, 9), dtype=np.uint8)
expected_square_5[2:7, 2:7] = 255
np.testing.assert_array_equal(
    dilatacao_binaria(large_image, square_5), expected_square_5
)

contract_image = with_noise.copy()
contract_element = square_3.copy()
for operation in (
    dilatacao_binaria,
    erosao_binaria,
    abertura_binaria,
    fechamento_binario,
):
    output = operation(contract_image, contract_element)
    assert_binary_output(output, contract_image.shape)
    np.testing.assert_array_equal(contract_image, with_noise)
    np.testing.assert_array_equal(contract_element, square_3)

print("Test passed!")
```
